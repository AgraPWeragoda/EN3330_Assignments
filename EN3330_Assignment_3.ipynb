Graph Generation for Question 1

```python
import numpy as np
import matplotlib.pyplot as plt


# Graph 1: Part (a)

plt.figure(figsize=(7, 5))

x = np.linspace(-1, 3, 400)
mus = [0, 1, 2, 3]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728']

for mu, col in zip(mus, colors):
    L = x**2 + mu * (1 - x)
    x_min = mu / 2.0
    L_min = mu - (mu**2) / 4.0
    plt.plot(x, L, label=f'$\\mu = {mu}$', color=col, lw=2.2)
    plt.plot(x_min, L_min, 'o', color=col, markersize=8)
    # Label the coordinates of each minimizer
    plt.annotate(f'({x_min:.1f}, {L_min:.2f})', 
                 xy=(x_min, L_min), xytext=(x_min - 0.25, L_min - 0.65),
                 fontsize=10, weight='bold', color=col)

plt.axvline(1.0, color='black', linestyle='--', alpha=0.7, label='Constraint boundary: $x = 1$')
plt.title(r'Lagrangian $L(\mu, x) = x^2 + \mu(1 - x)$ for Various $\mu$', fontsize=13, pad=12)
plt.xlabel('$x$', fontsize=12)
plt.ylabel(r'$L(\mu, x)$', fontsize=12)
plt.xlim(-1, 3)
plt.ylim(-1, 8)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=10)

plt.savefig("Q1_part_a_Lagrangian.png", dpi=300, bbox_inches='tight')
plt.show()

# Graph 2: Part (b) 

plt.figure(figsize=(7, 5))

mu_vals = np.linspace(0, 4, 300)
d_vals = mu_vals - (mu_vals**2) / 4.0

plt.plot(mu_vals, d_vals, color='#9467bd', lw=2.5, label=r'$d(\mu) = \mu - \frac{\mu^2}{4}$')
# Mark the dual optimal point
plt.plot(2.0, 1.0, 'r*', markersize=14, label=r'Dual Optimum: $(\mu^*=2, d^*=1)$')

plt.axhline(1.0, color='gray', linestyle=':', alpha=0.6)
plt.axvline(2.0, color='gray', linestyle=':', alpha=0.6)

plt.title(r'Dual Function $d(\mu) = \min_x L(\mu, x)$ vs $\mu$', fontsize=13, pad=12)
plt.xlabel(r'$\mu$', fontsize=12)
plt.ylabel(r'$d(\mu)$', fontsize=12)
plt.xlim(0, 4)
plt.ylim(-0.2, 1.2)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='lower center', fontsize=10)

plt.savefig("Q1_part_b_Dual_Function.png", dpi=300, bbox_inches='tight')
plt.show()

# ==============================================================================
# Graph 3: Part (c) - Primal Function g(x) vs x
# ==============================================================================
plt.figure(figsize=(7, 5))

x_feas = np.linspace(1.0, 3.0, 200)

# Feasible region where g(x) = x^2
plt.plot(x_feas, x_feas**2, color='#2ca02c', lw=3, label=r'$g(x) = x^2$ (Feasible region: $x \geq 1$)')

# Infeasible boundary representation (barrier wall to +infinity)
plt.axvline(1.0, color='red', lw=2.5, linestyle='-', label=r'$g(x) = +\infty$ (Infeasible region: $x < 1$)')

# Shaded background for the infeasible zone
plt.axvspan(-1.0, 1.0, color='red', alpha=0.08, label='Infeasible domain')

# Mark the constrained minimum
plt.plot(1.0, 1.0, 'ko', markersize=8, label=r'Optimal solution: $x^*=1, g(x^*)=1$')

plt.title(r'Primal Function $g(x) = \max_{\mu \geq 0} L(\mu, x)$', fontsize=13, pad=12)
plt.xlabel('$x$', fontsize=12)
plt.ylabel('$g(x)$', fontsize=12)
plt.xlim(-1, 3)
plt.ylim(0, 9)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', fontsize=10)

plt.savefig("Q1_part_c_Primal_Function.png", dpi=300, bbox_inches='tight')
plt.show()
```

Graph Generation for Question 2

```python
import numpy as np
import matplotlib.pyplot as plt


# Graph 1: 
plt.figure(figsize=(7, 4.5))

mu_vals = np.linspace(0, 5, 300)

L_mu = np.ones_like(mu_vals)

plt.plot(mu_vals, L_mu, color='#1f77b4', lw=2.5, label=r'$L(\mu, x^*) = 1$')

plt.plot(2.0, 1.0, 'ro', markersize=8, label=r'Saddle Point Value: $L(\mu^*, x^*) = 1$')

plt.axvline(2.0, color='gray', linestyle=':', alpha=0.7)
plt.title(r'Cross-Section 1: $L(\mu, x^*)$ vs $\mu$ (Fixed $x^* = 1$)', fontsize=12, pad=10)
plt.xlabel(r'$\mu$', fontsize=11)
plt.ylabel(r'$L(\mu, x^*)$', fontsize=11)
plt.xlim(0, 5)
plt.ylim(0, 3)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=10)

plt.savefig("Q2_part_iv_L_vs_mu.png", dpi=300, bbox_inches='tight')
plt.show()


# Graph 2

plt.figure(figsize=(7, 4.5))

x_vals = np.linspace(-1, 3, 400)

L_x = (x_vals - 1)**2 + 1

plt.plot(x_vals, L_x, color='#2ca02c', lw=2.5, label=r'$L(\mu^*, x) = (x - 1)^2 + 1$')

plt.plot(1.0, 1.0, 'ro', markersize=8, label=r'Minimizer: $(x^* = 1, L = 1)$')

plt.axvline(1.0, color='gray', linestyle=':', alpha=0.7)
plt.axhline(1.0, color='gray', linestyle=':', alpha=0.7)
plt.title(r'Cross-Section 2: $L(\mu^*, x)$ vs $x$ (Fixed $\mu^* = 2$)', fontsize=12, pad=10)
plt.xlabel('$x$', fontsize=11)
plt.ylabel(r'$L(\mu^*, x)$', fontsize=11)
plt.xlim(-1, 3)
plt.ylim(0, 6)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=10)

plt.savefig("Q2_part_iv_L_vs_x.png", dpi=300, bbox_inches='tight')
plt.show()
```

Graph Generation for Question 3

```python
import numpy as np
import matplotlib.pyplot as plt


plt.figure(figsize=(8, 5.5), layout='constrained')


mu_vals = np.linspace(0, 4.5, 400)


# For x = 0: L(mu, 0) = mu
line_x0 = mu_vals
# For x = 1: L(mu, 1) = 1
line_x1 = np.ones_like(mu_vals)
# For x = 2: L(mu, 2) = 4 - mu
line_x2 = 4 - mu_vals


lower_envelope_3lines = np.minimum(np.minimum(line_x0, line_x1), line_x2)


true_d_mu = mu_vals - (mu_vals**2) / 4.0

# Plot the individual lines
plt.plot(mu_vals, line_x0, '--', color='#1f77b4', lw=1.5, alpha=0.8, label=r'$L(\mu, 0) = \mu$')
plt.plot(mu_vals, line_x1, '--', color='#ff7f0e', lw=1.5, alpha=0.8, label=r'$L(\mu, 1) = 1$')
plt.plot(mu_vals, line_x2, '--', color='#2ca02c', lw=1.5, alpha=0.8, label=r'$L(\mu, 2) = 4 - \mu$')

# Highlight the lower envelope of the 3 lines
plt.plot(mu_vals, lower_envelope_3lines, color='black', lw=3, label='Lower Envelope of the 3 lines')

# Plot the true dual function
plt.plot(mu_vals, true_d_mu, color='#9467bd', lw=2.5, label=r'True $d(\mu) = \mu - \frac{\mu^2}{4}$')

# Formatting
plt.title(r'Dual Function as the Lower Envelope of Functions', fontsize=13, pad=12)
plt.xlabel(r'$\mu$', fontsize=12)
plt.ylabel(r'$L(\mu, x)$', fontsize=12)
plt.xlim(0, 4.5)
plt.ylim(0, 2.5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', fontsize=10)

# Save the figure
plt.savefig("Q3_part_b_Lower_Envelope.png", dpi=300)
plt.show()
```
